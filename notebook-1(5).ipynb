## Step 1 - File Access (2.5 points)

Write a Python function `accessData` that takes the dataset dictionary created in step 0 as an input and returns an extended dictionary including the following additions:

* Write code that accesses the dataset from its `resourceURL` using the python `requests` package:
 * detects whether it's and XML, CSV or JSON file by
     * checking whether the download URL **ends** with suffix "xml", "json", "csv" (in either upper- or lowercase)
     * checking whether the "Content-Type" HTTP header field contains information about the format, hinting on XML, JSON or CSV, i.e., check whether the substring XML, JSON or CSV appears in the "Content-Type" header in either upper- or lowercase. 
 * Detects the filesize from the HTTP header (converted to *Mebibyte*) of each data set, clearly documenting your actions (e.g. through commented code).

The result of the code below should extend your dictionaries `dataset1` and `dataset2` with two keys named 
* `"detectedFormat"` (which has one of the following values: `"XML"`, `"JSON"`, `"CSV"`, or `"unknown"`, if nothing could be detected from checking the suffix or HTTP header, or if the information in both was inconsistent)
* and `"filesizeMiB"` which contains the filesize in *Mebibyte* (MiB) from the number of bytes in the header-information, rounded to 8 decimal places. If there is no respective header information return 0.
* If the detected format is `"unknown"`, the expected filesize to be returned is also 0

````python
def accessData(datadict):
    # YOUR CODE HERE
    
     # 1. Get the download URL from the dataset dictionary.
    url = datadict["resourceURL"]

    # 2. Send an HTTP HEAD request to the URL.
    # HEAD returns information about the file (HTTP headers).
    # We do not need to download the entire file for this step.
    response = requests.head(url, allow_redirects=True)
    response.raise_for_status()

    # 3. Detect the format using the file extension.
    # Convert the URL to lowercase to ignore uppercase letters.
    url_lower = url.lower()

    url_format = ""

    if url_lower.endswith("csv"):
        url_format = "CSV"
    elif url_lower.endswith("json"):
        url_format = "JSON"
    elif url_lower.endswith("xml"):
        url_format = "XML"

    # 4. Detect the format using the Content-Type HTTP header.
    # For example, "application/json" indicates JSON.
    content_type = response.headers.get("Content-Type", "").lower()

    header_format = ""

    if "csv" in content_type:
        header_format = "CSV"
    elif "json" in content_type:
        header_format = "JSON"
    elif "xml" in content_type:
        header_format = "XML"

    # 5. Compare the results of both format checks.
    # If both formats disagree, the result is "unknown".
    if url_format and header_format and url_format != header_format:
        detected_format = "unknown"
    elif url_format:
        detected_format = url_format
    elif header_format:
        detected_format = header_format
    else:
        detected_format = "unknown"

    # 6. Read the filesize from the Content-Length HTTP header.
    # Content-Length contains the number of bytes.
    content_length = response.headers.get("Content-Length")

    # Default filesize is 0 if no size information is available.
    filesize = 0

    if detected_format != "unknown" and content_length is not None:

        # Convert bytes to MiB:
        # 1 MiB = 1024 * 1024 bytes.
        # Round the result to 8 decimal places.
        try:
            filesize = round(int(content_length) / (1024 * 1024), 8)
        except ValueError:
            filesize = 0

    # 7. Add the two new values to the dataset dictionary.
    datadict["detectedFormat"] = detected_format
    datadict["filesizeMiB"] = filesize

    # 8. Return the extended dictionary.
    return datadict

    
    
````

## Step 2  (2.5 points) - Format Validation

Establish that the two data files obtained are well-formed according to the detected data format (CSV, JSON, or XML). That is, the syntax used is valid according to accepted syntax definitions. Are there any violations of well-formedness?

Proceed as follows (for each data file, in turn): according to the "suspected" data format from Step 1:
Write a Python function `parseFile(datadict, format)` that that accesses the dataset from its `resourceURL`. The dataset should then be checked accordingly the given parser for the parameter `format` to check the following:

* CSV: Returns `True`, if a consistent delimiter out of `",",";","\t"` can be detected, such that each row has the same (> 1) number of elements, otherwise False
* JSON: Returns `True` if the file can be parsed with the `json` package, catching any parsing exceptions.
* XML: Returns `True` if the file can be parsed with the `xmltodict` package, catching any parsing exceptions.
* Returns `False` if any other format is supplied by the parameter.
     
In order to handle parsing exceptions and errors from the used packages, you can use [catching exceptions](https://docs.python.org/3/tutorial/errors.html), such that the program does not simply fail to check whether the file is parseable as the format specified in `format`    

````python
def parseFile(datadict, format):
    # YOUR CODE FOR STEP 2.2 HERE
    
     # 1. Check if the given format is supported.
    if format not in ["CSV", "JSON", "XML"]:
        return False

    try:
        # 2. Get the URL from the dataset dictionary.
        url = datadict["resourceURL"]

        # 3. Download the file using requests.
        response = requests.get(url)
        response.raise_for_status()

        # 4. Convert the downloaded content into text.
        text = response.content.decode("utf-8-sig")

        # 5. Check JSON files.
        if format == "JSON":

            # json.loads() checks if the JSON syntax is correct.
            json.loads(text)
            return True

        # 6. Check XML files.
        elif format == "XML":

            # xmltodict.parse() checks the XML structure.
            xmltodict.parse(text)
            return True

        # 7. Check CSV files.
        elif format == "CSV":

            # Try common CSV separators.
            delimiters = [",", ";", "\t"]

            for delimiter in delimiters:

                # Read the CSV text line by line.
                reader = csv.reader(
                    text.splitlines(),
                    delimiter=delimiter,
                    strict=True
                )

                rows = list(reader)

                # Skip empty files.
                if len(rows) == 0:
                    continue

                # Count the columns in the first row.
                column_count = len(rows[0])

                # A CSV file should have at least two columns.
                if column_count <= 1:
                    continue

                # Check whether all rows have the same number of columns.
                valid = True

                for row in rows:
                    if len(row) != column_count:
                        valid = False
                        break

                # If every row has the same number of columns,
                # the CSV structure is valid.
                if valid:
                    return True

            # No separator produced a valid CSV structure.
            return False

    # 8. Return False if downloading or parsing fails.
    except Exception:
        return False
    # YOUR CODE HERE
   
````

## Step 3 - Content analysis (4 points)

Similar to the Python function `parseFile(datadict,format)` above, now create a new Python function `describeFile(datadict)` that analyses the given file according to the respective format detected in Step 1 and returns a NEW dictionary containing the following information:

* for CSV files:
  * number of non-numeric columns,
    * ensure to try to transform any entry to a numeric type to ensure no number is accidentely seen as a string.
    * treat both, numbers with '.' as a decimal separator, as well as numbers with ',' as decimal separator as numeric.
    * ignore the header row, but otherwise a column counts as non-numeric, if it contains *any* non-numeric entry<br/><br/>
  * number of rows
    * for the number of rows *do* also count the header row (plus all data rows)!<br/><br/>
  * longest unique string: the character string with the most unique characters (any character counts, no cleaning needed, incl. the CSV header):
    * if there are no strings (i.e., non-numeric values) in your dataset, simply return "" as that value,
    * if there are several with the same maximum number of unique characters , return the first one occurring in the file.<br/><br/>
  * The resulting dictionary should have the following form:

    ```
    { "numberOfNonNumericColumns":  ...,
       "numberOfRows":  ... ,
       "longestUniqueString": ... }
    ```
    <br/>
* for JSON files: number of distinct attribute names, nesting depth, length of the longest dictionary appearing in an attribute value. That is, the resulting dictionary should have the following form:

    ```
    { "numberOfAttributes": ... ,
      "nestingDepth":  ... ,
      "longestDictLength": ... }
     ```
  <br/>Here:
  * `longestDictLength` should be set to 0 if no dictionary appears.
  * `nestingDepth` is defined as follows (Hint: you may want to look e.g. here for a start, [Nesting depth](https://www.tutorialspoint.com/find-depth-of-a-dictionary-in-python)):
    * a flat *list* of atomic values has depth 0,
    * a flat JSON *object* with only atomic attribute values has depth 1, 
    * a JSON attribute with another *object* as value (or another *object* as member of a list value!) increases the depth by 1
    * and so on.<br/>
  * `longestDictLength` should be self-explanatory, the length of the longest dictionry (anywhere) within the JSON file

* for XML files: number of different element and attribute a names (i.e. the sum of both), nesting depth, maximum float value in the dataset. That is, the resulting dictionary should have the following form:

    ```
    { "numberOfElementsAttributes": ... ,
      "nestingDepth":  ... ,
      "largestNumber" : ... }
     ```

  <br/>Here:

  * `numberOfElementsAttributes` is the number of distinct attributes and elements (Hint: after conversion using `xmltodict`) 
  * `nestingDepth` is defined as the nesting depth of elements.
  * `largestNumber`
    * it should be set to 0 if there are no numeric values present
    * only those values should be counted as numeric that are recognized as numeric by `xmltodict`, i.e. no conversion or alike needed as above in CSVs
  
For files that cannot be parsed with respective given format, the function should simply return an empty dictionary (`{}`).

##### Please ensure to keep the *names* and order of the elements/attributes as specified above! The test cases assume that your result has the same structure as the ones specified above. Basically this means for e.g. for a CSV that the first attribute should ALWAYS be the `"numberOfNonNumericColumns":`, etc.!

````python
def describeFile(datadict):
    # YOUR CODE HERE
    
     # Get the URL from the dataset dictionary.
    url = datadict.get("resourceURL", "")

    # Get the detected format from Step 1, if available.
    file_format = datadict.get("detectedFormat", "unknown")

    # If no format was detected, check the file extension.
    if file_format == "unknown":

        if url.lower().endswith(".csv"):
            file_format = "CSV"

        elif url.lower().endswith(".json"):
            file_format = "JSON"

        elif url.lower().endswith(".xml"):
            file_format = "XML"

    try:
        # Download the file and read its content.
        response = requests.get(url)
        response.raise_for_status()
        text = response.content.decode("utf-8-sig")

# CSV FILE
       
        if file_format == "CSV":
            try:
                separator = csv.Sniffer().sniff(
                    text[:1024], delimiters=",;\t"
                ).delimiter
            except csv.Error:
                separator = ","
        

            # Read the CSV file using the detected separator.
            rows = list(csv.reader(
                text.splitlines(), delimiter=separator))

            if len(rows) == 0:
                return {}

            # Count all rows, including the header.
            number_of_rows = len(rows)

            # This function checks if a value is numeric.
            def is_number(value):
                try:
                    # Commas and dots can both be decimal separators.
                    float(value.replace(",", "."))
                    return True
                except ValueError:
                    return False

            # Count columns containing non-numeric data.
            # The first row (header) is not included here.
            non_numeric_columns = 0

            for column in range(len(rows[0])):

                for row in rows[1:]:

                    if not is_number(row[column]):
                        non_numeric_columns += 1
                        break

            # Find the string with the most unique characters.
            # The header is included in this search.
            longest_string = ""
            most_unique_characters = 0

            for row in rows:
                for value in row:

                    if not is_number(value):
                        unique_characters = len(set(value))

                        if unique_characters > most_unique_characters:
                            most_unique_characters = unique_characters
                            longest_string = value

            return {
                "numberOfNonNumericColumns": non_numeric_columns,
                "numberOfRows": number_of_rows,
                "longestUniqueString": longest_string
            }

    
# JSON FILE

        elif file_format == "JSON":

            # Convert JSON text into Python objects.
            data = json.loads(text)

            # Store distinct attribute names.
            attribute_names = set()

            # Store the size of the largest dictionary.
            longest_dict = 0

            # Examine dictionaries and lists recursively.
            def examine_json(value):

                nonlocal longest_dict

                if isinstance(value, dict):

                    # Check the size of this dictionary.
                    if len(value) > longest_dict:
                        longest_dict = len(value)

                    # Examine all attributes and their values.
                    for key, item in value.items():
                        attribute_names.add(key)
                        examine_json(item)

                elif isinstance(value, list):

                    # Examine every item in the list.
                    for item in value:
                        examine_json(item)

            # Calculate how deeply objects are nested.
            def json_depth(value):

                if isinstance(value, dict):

                    max_depth = 0

                    for item in value.values():
                        depth = json_depth(item)

                        if depth > max_depth:
                            max_depth = depth

                    return max_depth + 1

                elif isinstance(value, list):

                    max_depth = 0

                    for item in value:
                        depth = json_depth(item)

                        if depth > max_depth:
                            max_depth = depth

                    return max_depth

                # Numbers, strings and other atomic values
                # do not increase nesting depth.
                return 0

            examine_json(data)

            return {
                "numberOfAttributes": len(attribute_names),
                "nestingDepth": json_depth(data),
                "longestDictLength": longest_dict
            }

# XML FILE

        elif file_format == "XML":

            # Convert XML into a Python dictionary.
            data = xmltodict.parse(text)

            # Store different element and attribute names.
            names = set()

            # Store the largest numeric value.
            largest_number = 0

            # Examine the XML structure recursively.
            def examine_xml(value, depth=0):

                nonlocal largest_number

                if isinstance(value, dict):

                    max_depth = depth

                    for key, item in value.items():

                        # Ignore the special key for text content.
                        if key != "#text":
                            names.add(key)

                        # Attributes start with @ and do not
                        # increase the element nesting depth.
                        if key.startswith("@") or key == "#text":
                            next_depth = depth
                        else:
                            next_depth = depth + 1

                        child_depth = examine_xml(item, next_depth)

                        if child_depth > max_depth:
                            max_depth = child_depth

                    return max_depth

                elif isinstance(value, list):

                    max_depth = depth

                    for item in value:
                        child_depth = examine_xml(item, depth)

                        if child_depth > max_depth:
                            max_depth = child_depth

                    return max_depth

                else:

                    # Do not convert strings into numbers.
                    # Only values already recognized as numeric count.
                    if isinstance(value, (int, float)) and not isinstance(value, bool):

                        if value > largest_number:
                            largest_number = value

                    return depth

            nesting_depth = examine_xml(data)

            return {
                "numberOfElementsAttributes": len(names),
                "nestingDepth": nesting_depth,
                "largestNumber": largest_number
            }

        # Return an empty dictionary for unsupported formats.
        return {}

    except Exception:
        # Return an empty dictionary if downloading or parsing fails.
        return {}
    
````
