## Step 1 - File Access (2.5 points)

Write a Python function `accessData` that takes the dataset dictionary created in step 0 as an input and returns an extended dictionary including the following additions:

* Write code that accesses the dataset from its `resourceURL` using the python `requests` package:
 * detects whether it's and XML, CSV or JSON file by
     * checking whether the download URL **ends** with suffix "xml", "json", "csv" (in either upper- or lowercase)
     * checking whether the "Content-Type" HTTP header field contains information about the format, hinting on XML, JSON or CSV, i.e., check whether the substring XML, JSON or CSV appears in the "Content-Type" header in either upper- or lowercase. 
 * Detects the filesize from the HTTP header (converted to *Mebibyte*) of each data set, clearly documenting your actions (e.g. through commented code).

The result of the code below should extend your dictionaries `dataset1` and `dataset2` with two keys named 
* `"detectedFormat"` (which has one of the following values: `"XML"`, `"JSON"`, `"CSV"`, or `"unknown"`, if nothing could be detected from checking the suffix or HTTP header, or if the information in both was inconsistent)
* and `"filesizeMiB"` which contains the filesize in *Mebibyte* (MiB) from the number of bytes in the header-information, rounded to 8 decimal places. If there is no respective header information return 0.
* If the detected format is `"unknown"`, the expected filesize to be returned is also 0

````python
def accessData(datadict):
     
    url = datadict["resourceURL"] #holt die URL des Datasets aus dem Dictionary

    response = requests.head(url, allow_redirects=True) #sendet eine HEAD Anfrage, um Informationen über die Datei zu erhalten. Mögliche Weiterleitungen werden berücksichtigt.
    response.raise_for_status() #prüft, ob die Anfrage erfolgreich war

    url_lower = url.lower() #wandelt die URL in kleinbuchstaben um

    url_format = "" #als erstes wird für das URL Format ein leerer String gesetzt.
    
    if url_lower.endswith("csv"): #prüfung, wie die URL endet.
        url_format = "CSV"

    if url_lower.endswith("json"):
        url_format = "JSON"

    if url_lower.endswith("xml"):
        url_format = "XML"

    content_type = response.headers.get("Content-Type", "").lower() #holt den Content Type aus den HTTP Headern und wandelt den in Kleinbuchstaben um.

    header_format = "" #auch für das header format wird als erstes ein leerer String gesetzt.

    if "csv" in content_type: #prüfung, welches dateiformat im Content Type angegeben ist.
        header_format = "CSV"
        
    if "json" in content_type:
        header_format = "JSON"
        
    if "xml" in content_type:
        header_format = "XML"

    if url_format and header_format and url_format != header_format: #verlgeicht die beiden ermittelten Dateiformate.
        detected_format = "unknown" #wenn beide vorhanden sind, aber unterschiedlich, wird das format als "unknown" gespeichert.
    elif url_format: #wenn ein Format über die URL gefunden wurde, wird dieses verwendet.
        detected_format = url_format
    elif header_format: #wenn ein format überden HTTP Header gefunden wurde, wird dieses verwendet.
        detected_format = header_format
    else:
        detected_format = "unknown" #wenn kein fromat gefunden wurde, wird "unknown" gespeichert.

   
    content_length = response.headers.get("Content-Length") #holt die Dateigröße in bytes aus dem HTTP Header
    filesize = 0 #fileseize ist als erstes 0

    if detected_format != "unknown" and content_length is not None: #wenn ein Format gefunden wurde und die Dateigröße nicht Null ist
        try:
            filesize = round(int(content_length) / (1024 * 1024), 8) #Berechene die Dateigröße von Bytes in Megabytes
        except ValueError:
            filesize = 0 #wenn die Umwandlung in eine ganze zahl nicht möglich ist, bleibt die Größe bei 0

    datadict["detectedFormat"] = detected_format #das erkannte Format wird zum Dictionary hinzugefügt.
    datadict["filesizeMiB"] = filesize #die berechnete Dateigröße kommt ebenfalls ins dictionary.

    return datadict #gibt das dictionary zurück.

print(accessData(dataset1)) 
print(accessData(dataset2))
````

## Step 2  (2.5 points) - Format Validation

Establish that the two data files obtained are well-formed according to the detected data format (CSV, JSON, or XML). That is, the syntax used is valid according to accepted syntax definitions. Are there any violations of well-formedness?

Proceed as follows (for each data file, in turn): according to the "suspected" data format from Step 1:
Write a Python function `parseFile(datadict, format)` that that accesses the dataset from its `resourceURL`. The dataset should then be checked accordingly the given parser for the parameter `format` to check the following:

* CSV: Returns `True`, if a consistent delimiter out of `",",";","\t"` can be detected, such that each row has the same (> 1) number of elements, otherwise False
* JSON: Returns `True` if the file can be parsed with the `json` package, catching any parsing exceptions.
* XML: Returns `True` if the file can be parsed with the `xmltodict` package, catching any parsing exceptions.
* Returns `False` if any other format is supplied by the parameter.
     
In order to handle parsing exceptions and errors from the used packages, you can use [catching exceptions](https://docs.python.org/3/tutorial/errors.html), such that the program does not simply fail to check whether the file is parseable as the format specified in `format`    

````python
def parseFile(datadict, format):
    
    if format not in ["CSV", "JSON", "XML"]: #dwenn das Format im Dictionary nicht als CSV, JSON oder XML detected wurde, gib False zurück
        return False

    try:
        url = datadict["resourceURL"] #holt die URL der DAtei aus dem Dictionary

        response = requests.get(url) #läst die Datei über die URL herunter
        response.raise_for_status() #auch hier überprüfen wir auf HTTP Fehler

        text = response.content.decode("utf-8-sig") #DAteiinhalt in Text umwandeln.

        if format == "JSON": #wenn format = Json, lese Text als JSON ein (Syntax)
            json.loads(text)
            return True #wenn syntax passt = True

        elif format == "XML": #ansonsten wenn format = xml, lese Text als XML ein (wieder Struktur überprüfung)
            xmltodict.parse(text)
            return True

        elif format == "CSV": #beim CSV müssen wir die Struktur manuell überprüfen
            delimiters = [",", ";", "\t"] #mögliche Trennzeichen sind Comma, Semicolon und Tabs

            for delimiter in delimiters: #jede Zeile des Files muss auf die möglichen Trennzeichen überprüft werden.
                reader = csv.reader(text.splitlines(), delimiter=delimiter, strict=True)

                rows = list(reader) #neue Liste, diesmal mit allen eingelesenen Zeilen

                if len(rows) == 0: #wenn die lieste leer, versuch das nächste Trennzeichen
                    continue

                column_count = len(rows[0]) #zählt die Anzahl der Spalten der ersten Zeile

                if column_count <= 1: #prüft ob die Datei mindestens zwei spalten hat
                    continue

                valid = True #setzt die überprüfung als erstes auf True

                for row in rows: #alle zeilen durchgehen
                    if len(row) != column_count: #wenn aktuelle Zeile nicht gleich viele spalten wie die erste hat = false
                        valid = False
                        break #beendet schleife, wenn kein CSV

                if valid: #wenn die Überprüfung erfolgreich war = True
                    return True
                
            return False #gibt flase zurück, wenn keines der Trennzeichen ein gültige Struktur ergibt
        
    except Exception: ##except fängt fehler beim herunterlade, dekodieren oder einlesen der DAtei ab. Wenn Fehler = False
        return False
    
print(parseFile(dataset1, "CSV"))
print(parseFile(dataset2, "JSON"))
````

Basically this means for e.g. for a CSV that the first attribute should ALWAYS be the `"numberOfNonNumericColumns":`, etc.!

````python
def describeFile(datadict):
    
    url = datadict.get("resourceURL", "") #hol den URL aus dem Datenset Dictionary, wenn kein url vorhande, bleibt der String leer

    file_format = datadict.get("detectedFormat", "unknown") #holt das format aus dem Dictionary, falls kein Format vorhanden, holt unknown

    if file_format == "unknown": #wenn kein Format erkannt wurde also format = unknown, dann überprüfe die URL auf Dateiendungen.

        if url.lower().endswith(".csv"):
            file_format = "CSV"

        elif url.lower().endswith(".json"):
            file_format = "JSON"

        elif url.lower().endswith(".xml"):
            file_format = "XML"

    try:
        response = requests.get(url) #versucht, die datei über die URL herunterzuladen
        response.raise_for_status()
        text = response.content.decode("utf-8-sig") #wandelt DAteiinhalt in String um
 
    # CSV Prüfung:
    
        if file_format == "CSV":
            try:
                separator = csv.Sniffer().sniff(text[:1024], delimiters=",;\t").delimiter #die CSV Sniffer Funktion versucht, die Trennzeichen automatisch zu erkennen
                
            except csv.Error: #wenn keine Trennzeichen erkannt, verwende comma als Trennzeichen.
                separator = ","
            
            rows = list(csv.reader(text.splitlines(), delimiter=separator)) #liest csv datei zeilenweise ein und speichert das als liste.

            if len(rows) == 0: #wenn keine Zeilen vorhande, kommt ein leeres Dictionary zurück
                return {}

            number_of_rows = len(rows) #zählt alle Reihen

            def is_number(value): #Funktion zur überprüfung von numerischen Werten
                try:
                    float(value.replace(",", ".")) #ersetzt commas durch Punkte und wandelt sie in einen String um.
                    return True
                except ValueError: #Wenn umwandlung nicht funktioniert, gib false zurcük
                    return False

            non_numeric_columns = 0 #zähler für spalten, welche keine numerischen werte enthalten
            
            for column in range(len(rows[0])): #geht alle spalten der Datei einzeln durch
                for row in rows[1:]: #alle spalten außer die erste, also den header, durchgehen
                    if not is_number(row[column]): #wenn die spalte eine nicht numerische zahl enthält, wird der zähler um eins erhöht
                        non_numeric_columns += 1
                        break

            longest_string = "" #nun zum längsten String
            most_unique_characters = 0 #string und zähler der unterschiedlichen zeichen im String auf leer bzw. null setzen

            for row in rows: #alle zeilen durchgehen
                for value in row: #alle werte der zeile durchgehen

                    if not is_number(value): #wenn der wert nicht numerisch ist, berechne die anzahl der unterschiedlichen zeichen und speicher den wert als variable
                        unique_characters = len(set(value))

                        if unique_characters > most_unique_characters: #wenn nun der wert dieser Variable unique characters größer als der bisher größte gespeicherte wert der variable most_unice_character ist
                            most_unique_characters = unique_characters #dann übernehme die Anzahl und speichere diesen wert als longest String
                            longest_string = value

            return { #gibt die Werte nun als dictionary zurück
                "numberOfNonNumericColumns": non_numeric_columns,
                "numberOfRows": number_of_rows,
                "longestUniqueString": longest_string
            }

    
    # JSON Prüfung

        elif file_format == "JSON": 

            data = json.loads(text) #wandlt den JSON-Text in Python Objekte um

            attribute_names = set() #erstellt ein leeres Set für alle unterschiedlichen Attributnamen

            longest_dict = 0 #setze größte anzahl an Einträgen auf null

            def examine_json(value): #funktion, um Json-Dicitionaries und Listen zu durchsuchen

                nonlocal longest_dict #ermöglicht die Veränderung der Variable aus der äußeren Funktion

                if isinstance(value, dict): #überprüft, ober der aktuelle wert ein dictionary ist

                    if len(value) > longest_dict: #hat dieses dicitonary mehr Einträge als das bisher größte
                        longest_dict = len(value)

                    for key, item in value.items(): #geht alle schlüssel und Werte des Dictionaries durch
                        attribute_names.add(key) #fügt den SChlüssel zum SEt der Attributnamen hinzu
                        examine_json(item) #durchsucht den zugehörigen WErt ebenfalls

                elif isinstance(value, list): #überprüft ob der aktuell wert eine liste ist
                    for item in value: #geht alle elemente der liste durch
                        examine_json(item) #durchsucht jedes Element der Liste ebenfalls

            
            def json_depth(value): #Funktion zur Berechnung der Verschachtelungstiefe

                if isinstance(value, dict): #wenn der aktuelle Wert ein Dictionary ist, setze tiefe auf 0
                    max_depth = 0 

                    for item in value.values(): #geht alle Werte des dicts durch
                        depth = json_depth(item) #berechnet verschachtelungstiefe des aktuellen werts

                        if depth > max_depth: #Wenn die aktuelle Tiefe größer als bisher größte Tiefe, setze größte Tiefe
                            max_depth = depth

                    return max_depth + 1 #gibt die größte Tiefe zurück und zählt das aktuelle dicts dazu

                elif isinstance(value, list): #gleiche überprüfung bei wenn der aktuelle wert eine liste ist
                    max_depth = 0

                    for item in value:
                        depth = json_depth(item)

                        if depth > max_depth:
                            max_depth = depth

                    return max_depth #gibt nur die größte tiefe zurück.
                
                return 0 #wenn keine Liste oder kein Dictionary dann 0

            examine_json(data) #starte die untersuche aller Daten im JSON

            return { #gibt die drei Ergebnisse zrück
                "numberOfAttributes": len(attribute_names),
                "nestingDepth": json_depth(data),
                "longestDictLength": longest_dict
            }

    #XML Prüfung

        elif file_format == "XML":

            data = xmltodict.parse(text) #wandelt den XML Text in eine Dictionary Struktur um
            names = set() #Erstelle leeres set für Attribut und Elementnamen.

            largest_number = 0 #setze den größten numerische Wert auf 0

            def examine_xml(value, depth=0): #funktion um XML Struktur zu untersuchen. Tiefe starte ab 0

                nonlocal largest_number #ermöglicht veränderung der Variable aus der äußeren Funktion

                if isinstance(value, dict): #wenn aktueller Wert ein Dictionary ist
                    max_depth = depth #speichert die bisher größte verschachtelungstiefe

                    for key, item in value.items(): #geht alle schlüssel und Werte des dictionaries durch 
                        if key != "#text": #wenn der schlüssel nicht für einen Textinhalt steht, füge Attributnamen zum SEt hinzu
                            names.add(key)

                        if key.startswith("@") or key == "#text": #überprüft ob es sich um ein Attribut oder Textinhalt handelt
                            next_depth = depth
                        else:
                            next_depth = depth + 1 #wenn nicht, dann erhöhe die Verschachtelungstiefe um 1

                        child_depth = examine_xml(item, next_depth) #untersucht zugehörigen wert und berechnet seine Tiefe

                        if child_depth > max_depth: #wenn diese Tiefe größer als die aktuell größte Tiefe ist, ersetze sie als gröte Tiefe
                            max_depth = child_depth

                    return max_depth #gib die größte gefundene Tiefe zurück

                elif isinstance(value, list): #das gleiche auch beim XML wieder für die Listen

                    max_depth = depth

                    for item in value:
                        child_depth = examine_xml(item, depth)

                        if child_depth > max_depth:
                            max_depth = child_depth

                    return max_depth

                else: #falls der aktuelle Wert keine Liste und kein Dictionary ist

                    if isinstance(value, (int, float)) and not isinstance(value, bool): #überprüfe, ob der Wert eine Zahl ist, aber kein Boolean

                        if value > largest_number: #überprüfe ob die Zahl größer als die bisher größte Zahl ist
                            largest_number = value

                    return depth #gibt die aktuelle Tiefe zurück

            nesting_depth = examine_xml(data) #startet die Untersuchung der XML Datie und speichert die größte Tiefe

            return { #gibt diie drei berechneten Attribute zurück
                "numberOfElementsAttributes": len(names),
                "nestingDepth": nesting_depth,
                "largestNumber": largest_number
            }

        return {} #gibt ein leeres Dictionary zuürck, wenn Format nicht unterstützt wird

    except Exception: #fängt fehler ab, bei fehler gib ebenfalls leeres dictionary zurück
        return {}
    
print(describeFile(dataset1))
print(describeFile(dataset2)) 
````
